Aquí está el código completo y corregido:

```python

from langchain_text_splitters import RecursiveCharacterTextSplitter

from pathlib import Path
from langchain_postgres.vectorstores import PGVector
from dotenv import load_dotenv
import os

# Configurar el modelo de embeddings

)

    texts=chunks,

    # metadatas=[{"fuente": "dummytext.txt"} for _ in chunks],
```

## Explicación del código:

1. **Carga y división de documentos**: Igual que tu código original.

2. **Creación de `PGVector`**: Al instanciar `PGVector`, automáticamente:
   - Crea la extensión `vector` de PostgreSQL si no existe
   - Crea las tablas necesarias en el esquema `langchain`:
     - `langchain_pg_collection`: Para almacenar colecciones
     - `langchain_pg_embedding`: Para almacenar los textos y embeddings

3. **`add_texts()`**: Este método:
   - Genera los embeddings automáticamente usando el modelo `OllamaEmbeddings`
   - Inserta tanto los textos como sus embeddings en la tabla `langchain_pg_embedding`
   - Asocia los documentos con la colección especificada (`"mi_coleccion"`)

## Tablas que se crean automáticamente:

1. **`langchain_pg_collection`**:
   - `uuid`: Identificador único de la colección
   - `name`: Nombre de la colección (en tu caso `"mi_coleccion"`)
   - `cmetadata`: Metadatos de la colección en formato JSON

2. **`langchain_pg_embedding`**:
   - `id`: Identificador único del documento
   - `collection_id`: Referencia a la colección
   - `embedding`: Vector de embeddings (tipo `vector` de pgvector)
   - `document`: El texto del chunk
   - `cmetadata`: Metadatos del documento en formato JSONB

## Si quieres usar embeddings precalculados:

Si ya tienes los embeddings calculados (como en tu código original donde tienes `embeddings = embeddings_model.embed_documents(chunks)`), puedes usar:

```python

# Usar el método from_embeddings

    text_embeddings=list(zip(chunks, embeddings)),
    embedding=embeddings_model,
    collection_name="mi_coleccion",
    connection=connection,
)

print("Base de datos creada con embeddings precalculados")
```

````python
from langchain_ollama import OllamaEmbeddings
from langchain_postgres.vectorstores import PGVector
from dotenv import load_dotenv
import os

# Cargar variables de entorno
load_dotenv()
SERVER_AI_URL = os.getenv("SERVER_AI_URL") 

# Configurar el modelo de embeddings (debe ser el mismo que usaste para crear)
model = "bge-m3:latest"
embeddings_model = OllamaEmbeddings(model=model, base_url=SERVER_AI_URL)

# Cadena de conexión (la misma que usaste para crear)
connection = "postgresql+psycopg://langchain:langchain@localhost:5432/langchain"

# Cargar el vector store existente
vector_store = PGVector.from_existing_index(
    embedding=embeddings_model,
    collection_name="mi_coleccion",  # Mismo nombre de colección
    connection=connection,
)

# Ahora puedes hacer búsquedas semánticas
query = "Who are Nonna Lucia and Giovanni Caruso?"

# Opción 1: Búsqueda simple con similitud
print("=== Búsqueda semántica simple ===")
results = vector_store.similarity_search(
    query=query,
    k=3  # Número de resultados a retornar
)

for i, doc in enumerate(results, 1):
    print(f"\nResultado {i}:")
    print(f"Contenido: {doc.page_content[:200]}...")  # Primeros 200 caracteres
    print(f"Metadatos: {doc.metadata}")

# Opción 2: Búsqueda con puntuación de similitud
print("\n=== Búsqueda con puntuación de similitud ===")
results_with_score = vector_store.similarity_search_with_score(
    query=query,
    k=3
)

for i, (doc, score) in enumerate(results_with_score, 1):
    print(f"\nResultado {i} (score: {score:.4f}):")
    print(f"Contenido: {doc.page_content[:200]}...")
    print(f"Metadatos: {doc.metadata}")

# Opción 3: Búsqueda con filtros por metadatos
print("\n=== Búsqueda con filtros ===")
results_filtered = vector_store.similarity_search(
    query=query,
    k=3,
    filter={"campo_metadata": "valor"}  # Si tienes metadatos en tus documentos
)

for i, doc in enumerate(results_filtered, 1):
    print(f"\nResultado {i}:")
    print(f"Contenido: {doc.page_content[:200]}...")
    print(f"Metadatos: {doc.metadata}")
    print("\n\n"*2)
````

## 2. Código completo con funcionalidades avanzadas

```python
from langchain_ollama import OllamaEmbeddings
from langchain_postgres.vectorstores import PGVector
from langchain.chains import RetrievalQA
from langchain.prompts import PromptTemplate
from dotenv import load_dotenv
import os

# Cargar variables de entorno
load_dotenv()
SERVER_AI_URL = os.getenv("SERVER_AI_URL")

class SemanticSearchEngine:
    def __init__(self, collection_name="mi_coleccion"):
        self.model = "bge-m3:latest"
        self.embeddings_model = OllamaEmbeddings(model=self.model, base_url=SERVER_AI_URL)
        self.connection = "postgresql+psycopg://langchain:langchain@localhost:5432/langchain"
        self.collection_name = collection_name
        
        # Cargar la base de datos vectorial
        self.vector_store = PGVector.from_existing_index(
            embedding=self.embeddings_model,
            collection_name=self.collection_name,
            connection=self.connection,
        )
    
    def simple_search(self, query, k=5):
        """Búsqueda semántica simple"""
        results = self.vector_store.similarity_search(query, k=k)
        return results
    
    def search_with_scores(self, query, k=5):
        """Búsqueda con puntuaciones de similitud"""
        results = self.vector_store.similarity_search_with_score(query, k=k)
        return results
    
    def search_with_threshold(self, query, k=10, score_threshold=0.7):
        """Búsqueda con umbral de similitud"""
        results = self.vector_store.similarity_search_with_score(query, k=k)
        # Filtrar por umbral
        filtered_results = [(doc, score) for doc, score in results if score <= score_threshold]
        return filtered_results
    
    def mmr_search(self, query, k=5, fetch_k=20, lambda_mult=0.5):
        """Búsqueda con Maximal Marginal Relevance (diversifica resultados)"""
        results = self.vector_store.max_marginal_relevance_search(
            query=query,
            k=k,
            fetch_k=fetch_k,
            lambda_mult=lambda_mult
        )
        return results
    
    def hybrid_search(self, query, k=5, filter_dict=None):
        """Búsqueda híbrida: semántica + filtros"""
        if filter_dict:
            results = self.vector_store.similarity_search(
                query=query,
                k=k,
                filter=filter_dict
            )
        else:
            results = self.vector_store.similarity_search(query, k=k)
        return results
    
    def get_by_ids(self, ids):
        """Recuperar documentos por sus IDs"""
        return self.vector_store.get_by_ids(ids)
    
    def get_collection_info(self):
        """Obtener información sobre la colección"""
        # Esta funcionalidad requiere una conexión directa a la BD
        import psycopg2
        
        conn = psycopg2.connect(
            dbname="langchain",
            user="langchain",
            password="langchain",
            host="localhost",
            port="5432"
        )
        
        cursor = conn.cursor()
        
        # Contar documentos
        cursor.execute("""
            SELECT COUNT(*) 
            FROM langchain_pg_embedding e
            JOIN langchain_pg_collection c ON e.collection_id = c.uuid
            WHERE c.name = %s
        """, (self.collection_name,))
        
        count = cursor.fetchone()[0]
        
        # Obtener ejemplo de documento
        cursor.execute("""
            SELECT e.document, LENGTH(e.document)
            FROM langchain_pg_embedding e
            JOIN langchain_pg_collection c ON e.collection_id = c.uuid
            WHERE c.name = %s
            LIMIT 1
        """, (self.collection_name,))
        
        example = cursor.fetchone()
        
        conn.close()
        
        return {
            "total_documents": count,
            "example_document_length": example[1] if example else 0
        }

# Uso de la clase
def main():
    # Inicializar el motor de búsqueda
    search_engine = SemanticSearchEngine(collection_name="mi_coleccion")
    
    # Obtener información de la colección
    info = search_engine.get_collection_info()
    print(f"Información de la colección:")
    print(f"- Total de documentos: {info['total_documents']}")
    print(f"- Longitud de ejemplo: {info['example_document_length']} caracteres\n")
    
    # Ejemplos de consultas
    queries = [
        "conceptos fundamentales de machine learning",
        "aplicaciones de inteligencia artificial",
        "historia de la computación"
    ]
    
    for query in queries:
        print(f"\n{'='*60}")
        print(f"Consulta: '{query}'")
        print('='*60)
        
        # Búsqueda simple
        print("\n1. Búsqueda simple (top 3):")
        results = search_engine.simple_search(query, k=3)
        for i, doc in enumerate(results, 1):
            print(f"\n   Resultado {i}:")
            content_preview = doc.page_content[:150] + "..." if len(doc.page_content) > 150 else doc.page_content
            print(f"   {content_preview}")
        
        # Búsqueda con scores
        print("\n2. Búsqueda con puntuaciones:")
        results_with_scores = search_engine.search_with_scores(query, k=3)
        for i, (doc, score) in enumerate(results_with_scores, 1):
            print(f"\n   Resultado {i} (score: {score:.4f}):")
            content_preview = doc.page_content[:100] + "..." if len(doc.page_content) > 100 else doc.page_content
            print(f"   {content_preview}")

if __name__ == "__main__":
    main()
```

## 3. Sistema de preguntas y respuestas (Q&A) usando la base de datos

```python
from langchain_ollama import OllamaEmbeddings
from langchain_postgres.vectorstores import PGVector
from langchain.chains import RetrievalQA
from langchain.prompts import PromptTemplate
from langchain_ollama import OllamaLLM
from dotenv import load_dotenv
import os

load_dotenv()

class QASystem:
    def __init__(self, llm_model="llama3.2:latest"):
        # Configurar embeddings
        self.embeddings = OllamaEmbeddings(
            model="bge-m3:latest",
            base_url=os.getenv("SERVER_AI_URL")
        )
        
        # Configurar LLM para respuestas
        self.llm = OllamaLLM(
            model=llm_model,
            base_url=os.getenv("SERVER_AI_URL"),
            temperature=0.3
        )
        
        # Cargar vector store
        self.vector_store = PGVector.from_existing_index(
            embedding=self.embeddings,
            collection_name="mi_coleccion",
            connection="postgresql+psycopg://langchain:langchain@localhost:5432/langchain",
        )
        
        # Configurar retriever
        self.retriever = self.vector_store.as_retriever(
            search_type="similarity",
            search_kwargs={"k": 4}
        )
        
        # Plantilla personalizada para respuestas
        self.prompt_template = """Eres un asistente útil que responde preguntas basándose en el contexto proporcionado.

        Contexto:
        {context}

        Pregunta: {question}

        Instrucciones:
        1. Responde únicamente basándote en el contexto proporcionado
        2. Si la respuesta no está en el contexto, di "No tengo información suficiente sobre eso en los documentos"
        3. Sé conciso y claro
        4. Cita los fragmentos relevantes del contexto cuando sea apropiado

        Respuesta:"""
        
        self.prompt = PromptTemplate(
            template=self.prompt_template,
            input_variables=["context", "question"]
        )
        
        # Crear la cadena de Q&A
        self.qa_chain = RetrievalQA.from_chain_type(
            llm=self.llm,
            chain_type="stuff",
            retriever=self.retriever,
            return_source_documents=True,
            chain_type_kwargs={"prompt": self.prompt}
        )
    
    def ask_question(self, question):
        """Hacer una pregunta al sistema"""
        print(f"\nPregunta: {question}")
        print("-" * 50)
        
        # Obtener respuesta
        result = self.qa_chain.invoke({"query": question})
        
        # Mostrar respuesta
        print(f"Respuesta: {result['result']}")
        
        # Mostrar fuentes (opcional)
        print("\nFuentes utilizadas:")
        for i, doc in enumerate(result['source_documents'], 1):
            print(f"\nFuente {i}:")
            print(f"Contenido: {doc.page_content[:200]}...")
            if doc.metadata:
                print(f"Metadatos: {doc.metadata}")
        
        return result

# Uso del sistema Q&A
def main():
    # Inicializar sistema
    qa_system = QASystem(llm_model="llama3.2:latest")
    
    # Ejemplos de preguntas
    questions = [
        "¿Qué es machine learning y cómo funciona?",
        "¿Cuáles son las principales aplicaciones de la IA?",
        "Explica los conceptos clave de deep learning"
    ]
    
    for question in questions:
        qa_system.ask_question(question)
        print("\n" + "="*80 + "\n")

if __name__ == "__main__":
    main()
```

```python
        LIMIT 3
    """)
    
    examples = cursor.fetchall()
    print(tabulate(examples,
                   headers=["ID", "Contenido (primeros 100 chars)", "Metadatos"],
                   tablefmt="grid"))

        ORDER BY dimensiones
    """)
    
    dimensions = cursor.fetchall()
    print(tabulate(dimensions,
                   headers=["Dimensiones", "Cantidad"],
                   tablefmt="grid"))
    
    conn.close()

if __name__ == "__main__":
    explore_database()
```

## Resumen de funcionalidades:

1. **Cargar DB existente**: `PGVector.from_existing_index()`
2. **Búsqueda simple**: `similarity_search()`
3. **Búsqueda con scores**: `similarity_search_with_score()`
4. **Búsqueda MMR**: `max_marginal_relevance_search()` (para diversidad)
5. **Búsqueda con filtros**: Pasar parámetro `filter`
6. **Sistema Q&A**: Combinar con `RetrievalQA` y un LLM
7. **Explorar DB**: Conectar directamente con `psycopg2` para análisis

El sistema buscará automáticamente los chunks de texto más similares a tu consulta usando la similitud coseno (por defecto) entre los embeddings de tu query y los embeddings almacenados en la base de datos.
